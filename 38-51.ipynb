# #50 Reverse an Array Without Extra Space 

```python
num = [1,2,3,4,5,6,7,8,9]
def rev_array(num):
    if not isinstance(num,list):
        raise TypeError("input must be list")
    left = 0
    right = len(num)-1
    while left < right:
        num[left],num[right] = num[right],num[left]
        left += 1
        right -= 1
    return num
if __name__ == "__main__":
    try:
        user_input = [1,2,3,4,5,6,7,8,9]
        result = rev_array(user_input)
        print(result)
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
    
```

# # 49 Armstrong Number in Python

```python
def is_armstrong(number):
    temp_copy = number
    digit_count = 0
    while temp_copy > 0:
        digit_count = digit_count + 1
        temp_copy = temp_copy // 10

    total = 0
    temp_copy = number
    while temp_copy > 0:
        digit = temp_copy % 10
        total = total + digit ** digit_count
        temp_copy = temp_copy // 10
    return total == number

if __name__ == "__main__":
    try:
        user_input = input("Enter your value")
        user_input = int(user_input)
        result = is_armstrong(user_input)
        print(result)
    except Exception as e:
        print(e)
        
    
```

# #47 Find First Non-Repeating Character

```python

def first_char(text):
    
    char_count = {}
    for i in text:
        if i in char_count:
            char_count[i] += 1
        else:
            char_count[i] = 1

    for i in text:
        if char_count[i] == 1:
            return i

if __name__ == "__main__":
    text = "swwIss"
    try:
        result = first_char(text)
        print(f"First Non-Repeating Character : {result}")
    except Exception as e:
        print(e)
```

# #46 Remove Nth occurrence of the character from a string

```python
text = "these"
def remove_occurrence(text,remove_char,n):
    count = 0
    list1 = []
    for i in text:
        if i == remove_char:
            count = count + 1
            if count == n:
                continue
        list1.append(i)
    return "".join(list1)

if __name__ == "__main__":
    try:
        text = "these"
        remove = "e"
        num = 2
        result = remove_occurrence(text,remove,num)
        print(result)
    except Exception as e:
        print(e)
```

# #43 Convert Binary to Decimal

```python
def binary_decimal(text):
    total = 0
    power = 0
    for i in reversed(text):
        if i == "1":
            total = total + 2**power
        power = power + 1
    return total
            
```

```python
num = [2,4,5,6,7]
num.sort()
count = len(num)
for i in num:
    x = count//2
print(num[x])
```

```python
def find_median(num):
    num.sort()
    length = len(num)
    mid_val = length//2
    if length%2 != 0:
        return num[mid_val]
    else:
        return num[mid_val-1] + num[mid_val/2]

if __name__ == "__main__":
    try:
        num = [2,4,5,6,7]
        result = find_median(num)
        print(result)
    except Exception as e:
        print(e)
```

# #39 Implement Queue in Python using List

```python
class Queue:
    def __init__(self):
        self.queue = []
    def is_empty(self):
        return len(self.queue) == 0
          
    def enqueue(self,value):
        self.queue.append(value)
        print(f"{value} enqueue, queue {self.queue}")
    def dequeue(self):
        if self.is_empty():
            print(f"queue is empty : {self.queue}")
        else:
            removed = self.queue.pop(0)
            print(f"{removed} is dedequeue, queue {self.queue}")
    def peek(self):
        if self.is_empty():
            print("queue is empty")
        else:
            print(f"queue {self.queue[0]}")

q = Queue()
        
```

# #38 Convert List of Tuples to Dictionary

```python
data = [("a",1),("b",3),("c",4),("s",2)]
```

```python
def list_to_dict(data):
    my_dict = {}
    for key,value in data:
        my_dict[key] = value
    return my_dict

if __name__ == "__main__":
    try:
        result = list_to_dict(data)
        print(result)
    except Exception as e:
        print(e)
```

```python
def reverse_in_place(arr):
    left, right = 0, len(arr) - 1
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr  # same list object, just for convenience
```

```python
reverse_in_place([10,20,30,40,50])
```

```python
def find_first_duplicate(lst):
    seen = []
    for item in lst:
        if item in seen:
            return item
        seen.append(item)
    return None

print(find_first_duplicate([10, 5, 3, 4, 3, 5, 6]))  # 3
```
