```python
def is_anagram(str1:str, str2: str):
    if not isinstance(str1,str) or not isinstance(str2,str):
        raise TypeError("only string value")
    str1 = str1.replace(" ","").lower()
    str2 = str2.replace(" ","").lower()
    count = {}
    if len(str1) != len(str2):
        return False
    for i in str1:
        if i in count:
            count[i] += 1
        else:
            count[i] = 1
    for i in str2:
        if i not in count:
            return False
            
        if count[i] < 0:
            return False
        count[i] -= 1
    else:
        return True
if __name__ == "__main__":
    str1 = "abc"
    str2 = "bca"
    try:
        result = is_anagram(str1,str2)
        if result:
            print("this is anagram")
        else:
            print("this is not an anagram")
    except TypeError as e:
        print(e)
    except Exception as e:
        print(e)
    
```

```python
def cal_power(base_value, power_value):
    if not isinstance(base_value, int) or not isinstance(power_value, int):
        raise TypeError("Inputs must be integers")
    if base_value < 0 or power_value < 0:
        raise ValueError("Inputs must be greater than or equal to 0")
    total = 1
    for i in range(power_value):
        total = total * base_value
    return total

if __name__ == "__main__":
    input1 = input("Enter base value: ")
    input2 = input("Enter power value: ")
    try:
        input1 = int(input1)
        input2 = int(input2)
        result = cal_power(input1, input2)
        print(f"{input1} power of {input2} = {result}")
    except (ValueError, TypeError) as e:
        print(e)
    except Exception as e:
        print(e)
```

# #32  Reverse an Integer

```python
def rev_int(num):
    revesed = 0
    sign = -1 if num < 0 else 1
    num = abs(num)
    while num > 0:
        digit = num % 10
        revesed = revesed * 10 + digit
        num = num // 10
    return sign*revesed
if __name__ == "__main__":
    try:
        user_input = -12345
        result = rev_int(user_input)
        print(f" Reverse an Integer : {result}")
    except Exception as e:
        print(e)
```

# #33 Implement Stack Using List in Python (LIFO Explained) 

```python
class Stack:
    def __init__(self):
        self.stack = []
    def is_empty(self):
        return len(self.stack) == 0
    def push(self,value):
        self.stack.append(value)
        print(f"{value} is pushed")
    def pop(self):
        if self.is_empty():
            print("stack is empty")
            return None
        return self.stack.pop()
    def peek(self):
        if self.is_empty():
            print("stack is empty")
            return None
        return self.stack[-1]
    def __str__(self):
        return str(self.stack)

if __name__ == "__main__":
    obj = Stack()
    obj.push(10)
    obj.push(20)
    obj.push(30)
    print("peek : ",obj.peek())
    print("pop : ",obj.pop())
    print("pop : ",obj.pop())
    print("pop : ",obj.pop())
    
```

# #34 Find Longest Word in a Sentence

```python
text = """This is a very common Python coding interview question asked in companies like TCS, Infosys, Wipro, and even product-based companies.
In this video, you will learn:"""
```

```python
def longest_word_finder(text):
    import string
    text = text.lower()
    pattern = str.maketrans("","",string.punctuation)
    clean_text = text.translate(pattern)
    
    word_list = clean_text.split()
    word_count = {}
    
    for i in word_list:
        if i not in word_count:
            word_count[i] = len(i)
            
    max_len = max(word_count.values())
    longest_word = [key for key,value in word_count.items() if value == max_len]
    
    return longest_word


if __name__ == "__main__":
    try:
        user_input = text
        result = longest_word_finder(user_input)
        print(result)
    except Exception as e:
        print(e)
```

# #35 Check Power of 2

```python
def power_check(num):
    if num <= 0:
        return False,None
    power = 0
    while num%2 == 0:
        num = num//2
        power += 1
    if num == 1:
        return True,power
    else:
        return False,None

if __name__ == "__main__":
    try:
        user_input = 16
        result = power_check(user_input)
        print(result)
    except Exception as e:
        print(e)
```

# #36  Flatten a Nested List in Python |

```python
list1 = [2,[34,4],[23],6]
```

```python
def flatten(list1):
    result = []
    temp_list = list(list1)
    while temp_list:
        item = temp_list.pop()
        if isinstance(item,list):
            temp_list.extend(item)
        else:
            result.append(item)
    return result[::-1]

if __name__ == "__main__":
    try:
        user_input = [2,[34,4],[23],6]
        final = flatten(user_input)
        print(final)
    except Exception as e:
        print(e)
```

# #37 Find Pairs with Given Sum in a List 

```python
def find_pairs(num):
    target = 7
    pairs = []
    for i in range(len(num)):
        for j in range(i+1,len(num)):
            if num[i] + num[j] == target:
                pairs.append((num[i],num[j]))
    return pairs

if __name__ == "__main__":
    try:
        user_input = [1,2,3,4,5,6]
        result = find_pairs(user_input)
        print(result)
    except Exception as e:
        print(e)
        
```
